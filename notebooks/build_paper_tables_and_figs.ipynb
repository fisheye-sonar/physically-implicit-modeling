# The paper's tables and figures — the replication notebook

This notebook reproduces every table and figure in the paper from the scored runs, and ships with the public
codebase for that purpose. It reads each listed run's `scores.json` (written by `master_eval.ipynb`) and the floors
under `runs/_baselines/` — nothing under `experiments/`. All selection and drawing live in
`pim.figures.tables`; this notebook sets the run lists, the basis and the display knobs, and calls one function per
table or figure. Every table is an image; panels are drawn WITHOUT titles here (the paper's captions carry them —
`T.SHOW_TITLES = False`). The appendix's tables are in `build_appendix_tables_and_figs.ipynb` (predictive loss beside
the Bayes floor — it replaced the old "Table 4" cell here, 2026-09-19). The long list of every run and target is
`build_full_tables.ipynb`; the figures here are tailored for the paper and may drift from it.

Conventions: Othello above discworld with a heavy rule between them; rows follow the run lists; Othello's Edit Index
is the **symmetric-difference** construction (the union stays in `scores.json`); every per-cell decodability value
is that cell's own optimum over residual points; probes are held out by sequence; the editors shown are **PI, GS,
IM** (ND and IM-NN remain in `scores.json`).


```python
# [1] THE RUN LISTS — the only thing to edit. Othello first, then discworld; the order here is the row order.
RUNS_OTH = ['L-oth-20m', 'L-oth-adjacent-flip-20m', 'L-oth-adjacent-20m', 'L-oth-noflip-20m']
RUNS_DW = ['L-dw-noiseless-20m', 'L-dw-blink-20m', 'L-dw-128ray-20m', 'L-dw-16ray-20m', 'L-dw-8ray-20m', 'L-dw-5ray-20m']

import sys
from pathlib import Path

import matplotlib.pyplot as plt

REPO = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(REPO))
from pim.figures import tables as T

# THE BASIS (2026-09-15, Sevan): every discworld number in these tables — decodability, floors, PI / GS / IM —
# comes from this regression basis and none from the other. "frustum" (canonical) or "cartesian" (every run
# carries both blocks). A run without the requested block (dw-8ray-obs5: cartesian only) shows in the block it
# has, marked * when frustum is requested. Grid-defined targets are the same in either basis.
BASIS = 'cartesian'
T.set_basis(BASIS)

# TITLES (2026-09-15, Sevan): none on the panels in THIS notebook — the paper's captions carry them.
T.SHOW_TITLES = False

# Seed replicates (<run>__seed<k>) fold into ± columns. Guard: pooled only at a MATCHED training budget
# (±10%); set pool_budgets=True to pool every budget of a run (Table 5 then says 'mixed budgets').
POOL_BUDGETS = False
F = T.collect(RUNS_OTH, RUNS_DW, pool_budgets=POOL_BUDGETS, label='paper')
print(f'{len(F.df)} (run, target) rows · frame-set runs {sorted(F.frame_set) or "none"} · '
      f'replicate cells {len(F.rep_sd)}' + (f' · MISSING {F.missing}' if F.missing else ''))
```

```python
# [2] TABLE 1 — decodability against its floors (right-aligned observation; random-init of the listed architectures)
fig = T.table_decodability(F, '1')
plt.show() if fig else print('Table 1: nothing to draw')
```

```python
# [2b] TABLE 1b — the inverse map's held-out R² beside the nearest-state retrieval's, per run
#      (each at the point the IM editor's best arm writes, and at its own best point)
fig = T.table_inverse_r2(F, '1b')
plt.show() if fig else print('Table 1b: nothing to draw')
```

```python
# [3] TABLE 2 — editability (a) Edit Index with the unedited floor, (b) fidelity ratio; TABLE 2b the arms
fig = T.table_editability(F, '2')
plt.show() if fig else print('Table 2: nothing to draw')
# fig = T.table_arms(F, '2b')
# plt.show() if fig else None
```

```python
# [4] TABLE 2c — the gridified discworld targets only, coarse → fine
fig = T.table_gridified(F, '3')
plt.show() if fig else print('Table 2c: no gridified targets among the listed runs')
```
